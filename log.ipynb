```python
import matplotlib.pyplot as plt
```

```python
LOG_FILE = "outputs/log/log_run_5195355.txt"
```

```python
with open(LOG_FILE, "r") as file:
    lines = file.readlines()

lines
```

```python
overalls = [float(line.split("=")[-1]) for line in lines if "Overall" in line]
# overalls
overalls
```

```python
max(overalls)
```

```python
epochs = range(1, 50, 3)  # Number of epochs corresponds to the length of the overalls list
list(epochs)
```

```python

# Plotting
plt.plot(epochs, overalls, marker='o', linestyle='-', color='b')

# Labels and title
plt.xlabel('Epochs')
plt.ylabel('Correct Answers')
plt.title('Correct Answers per Epoch')
# plt.ylim(top=10)
# Show grid for better readability
plt.grid(True)

# Show plot
plt.show()
```

```python
losses = [float(line[16:].split()[0].replace(",","")) for line in lines if "Epoch:" in line]
losses
```

```python
epochs = range(1, len(losses) + 1)  # Number of epochs corresponds to the length of the overalls list

# Plotting
plt.plot(epochs, losses, marker='o', linestyle='-', color='orange')

# Labels and title
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss')

# Show grid for better readability
plt.grid(True)

# Show plot
plt.show()
```
